```python
%%writefile app.py
import streamlit as st
import sqlite3
import joblib
import pandas as pd

# 1. Carregar o modelo treinado
try:
    model = joblib.load('modelo_biomedicina.pkl')
except:
    st.error("Erro: Arquivo 'modelo_biomedicina.pkl' não encontrado.")

# --- FUNÇÕES DE BANCO DE DADOS ---
def validar_login(user, password):
    conn = sqlite3.connect('clinica.db')
    cursor = conn.cursor()
    cursor.execute('SELECT * FROM usuarios WHERE user=? AND password=?', (user, password))
    res = cursor.fetchone()
    conn.close()
    return res

def cadastrar_novo_usuario(user, password):
    conn = sqlite3.connect('clinica.db')
    cursor = conn.cursor()
    try:
        cursor.execute('INSERT INTO usuarios (user, password) VALUES (?, ?)', (user, password))
        conn.commit()
        return True
    except:
        return False
    finally:
        conn.close()

# --- CONTROLE DE SESSÃO ---
if 'logado' not in st.session_state:
    st.session_state['logado'] = False

# --- INTERFACE ---
st.sidebar.title("Menu Principal")

if not st.session_state['logado']:
    menu = st.sidebar.selectbox("Navegação", ["Login", "Criar Nova Conta"])
else:
    menu = st.sidebar.selectbox("Navegação", ["Dashboard", "Cadastro de Pacientes", "Lista de Pacientes", "Lançamento de Resultados", "Avaliação do Modelo", "Sair"])

# --- LÓGICA DAS TELAS ---
if menu == "Login":
    st.title("Acesso ao Sistema")
    u = st.text_input("Usuário/E-mail")
    p = st.text_input("Senha", type="password")
    if st.button("Entrar"):
        if validar_login(u, p):
            st.session_state['logado'] = True
            st.rerun()
        else:
            st.error("Usuário ou senha incorretos.")

elif menu == "Criar Nova Conta":
    st.title("Cadastro de Novo Usuário")
    new_user = st.text_input("Defina seu Usuário ou E-mail")
    new_pass = st.text_input("Defina sua Senha", type="password")
    confirm_pass = st.text_input("Confirme sua Senha", type="password")
    if st.button("Finalizar Cadastro"):
        if new_pass == confirm_pass and new_user:
            if cadastrar_novo_usuario(new_user, new_pass):
                st.success("Conta criada com sucesso!")
            else:
                st.error("Erro ao criar conta.")
        else:
            st.error("As senhas não coincidem.")

elif st.session_state['logado']:
    if menu == "Dashboard":
        st.title("Bem-vindo ao Sistema Clínico")
        st.info("Sistema integrado: Banco de Dados + IA + Interface")

    elif menu == "Cadastro de Pacientes":
        st.title("Cadastrar Novo Paciente")
        nome_paciente = st.text_input("Nome Completo")
        idade_paciente = st.number_input("Idade", min_value=0)
        if st.button("Salvar no Banco SQL"):
            conn = sqlite3.connect('clinica.db')
            conn.execute('INSERT INTO pacientes (nome, idade) VALUES (?, ?)', (nome_paciente, idade_paciente))
            conn.commit()
            conn.close()
            st.success(f"Paciente '{nome_paciente}' salvo com sucesso!")

    elif menu == "Lista de Pacientes":
        st.title("Pacientes Registrados")
        conn = sqlite3.connect('clinica.db')
        df_p = pd.read_sql_query("SELECT * FROM pacientes", conn)
        conn.close()
        st.dataframe(df_p, use_container_width=True)

    elif menu == "Lançamento de Resultados":
        st.title("Análise de Risco com IA")
        g = st.number_input("Glicose")
        pr = st.number_input("Pressão")
        i = st.number_input("IMC")
        c = st.number_input("Colesterol")
        if st.button("Processar Diagnóstico"):
            dados = pd.DataFrame([[g, pr, i, c]], columns=['glicose', 'pressao', 'imc', 'colesterol'])
            pred = model.predict(dados)
            if pred[0] == 1:
                st.error("### ⚠️ Grau de Risco: ALERTA")
            else:
                st.success("### ✅ Grau de Risco: NORMAL")

    elif menu == "Avaliação do Modelo":
        st.title("Métricas do Modelo")
        st.write("Acurácia: 94% | Precisão: 92%")

    elif menu == "Sair":
        st.session_state['logado'] = False
        st.rerun()
```
